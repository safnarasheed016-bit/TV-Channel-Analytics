```python

numeric_columns = [
    "Avg_Viewers",
    "Avg_Rating",
    "Total_Likes",
    "YouTube_Views",
    "Social_Media_Followers",
    "Monthly_Reach",
    "Subscription_Fee",
    "Years_Active"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
numeric_columns = df.select_dtypes(include="number").columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = df.select_dtypes(include="str").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df = df.drop_duplicates()
```

```python
df.loc[(df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5), "Avg_Rating"] = df["Avg_Rating"].median()
```

```python
df.loc[df["Subscription_Fee"] < 0, "Subscription_Fee"] = df["Subscription_Fee"].median()
```

```python
df.loc[df["Years_Active"] < 0, "Years_Active"] = df["Years_Active"].median()
```

```python
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df = df.replace(
    ["NA", "N/A", "na", "n/a", "unknown", "Unknown", ""],
    pd.NA
)
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.to_csv("cleaned_tv_channel_dataset.csv", index=False)
```
